### Introduction to HTTP

```python
import requests
import json

urlraw = "https://raw.githubusercontent.com/andrewbeattycourseware/wsaa-courseware/refs/heads/main/code/Topic-01-introduction/book.json"
response = requests.get(urlraw)
data = response.json()
print(json.dumps(data, indent=4))
```

```python
#Encode

import urllib.parse
query = "Python programming"
encoded_query = urllib.parse.quote(query)
print(encoded_query)

params = {'q': encoded_query, 'as_sitesearch': 'www.urlencoder.io'}
parsedparams = urllib.parse.urlencode(params)
print(parsedparams)
```

### Train Time

```python
from xml.dom.minidom import parseString
import requests
import csv

retrievetags = ['TrainStatus',
            'TrainLatitude',
            'TrainLongitude',
            'TrainCode',
            'TrainDate',
            'PublicMessage',
            'Direction'
            ]
url = "http://api.irishrail.ie/realtime/realtime.asmx/getCurrentTrainsXML"
page = requests.get(url)

doc = parseString(page.content)
print(doc.toprettyxml(newl=''))

with open('trainxml.xml', 'w') as xmlfp:doc.writexml(xmlfp)
```
