```python
import requests

url1 = "http://atu.ie"

response = requests.get(url1)
print(response.status_code)
```

```python
url2 = "https://andrewbeatty1.pythonanywhere.com/"

urlbook = url2 + "books"
response = requests.get(urlbook)
print(response.status_code)
print(response.json())
```

```python
import csv
import json

if response.status_code == 200:
    books_data = response.json()
    bookscsv = "books.csv"
    if books_data:
        heders = books_data[0].keys()
        with open(bookscsv, 'w', newline='', encoding='utf-8') as csvfile:
            writer = csv.DictWriter(csvfile, fieldnames=heders)
            writer.writeheader()
            writer.writerows(books_data)
        print(f"Data has been written to {bookscsv}")
        print(f"Total records: {len(books_data)}")
    else:
        print("No data found in the response.")
else:
    print(f"Failed to retrieve data. Status code: {response.status_code}")
```
