```python
import requests
import json

url = "http://andrewbeatty1.pythonanywhere.com/books"

def getAllBooks():
    reponse = requests.get(url)
    return reponse.json()

def getBookById(book_id):
    geturl = url + "/" + str(book_id)
    reponse1 = requests.get(geturl)
    return reponse1.json()

def createBook(book):
    reponse2 = requests.post(url, json=book)
    return reponse2.json()

def updateBook(book_id, book):
    updateurl = url + "/" + str(book_id)
    reponse3 = requests.put(updateurl, json=book)
    return reponse3.json()

print(getAllBooks())
print(getBookById(1666))
print(createBook({
    'author': 'Jane Smith',
    'title': 'New Book',
    'price': '19.99'
}))
```
